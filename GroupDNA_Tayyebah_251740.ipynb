# Feature 1 — Chat Parser

The parser reads the WhatsApp export line by line and identifies
real messages, system messages, media messages, deleted messages,
and multi-line messages.

```python
def looks_like_message_line(line):
    if len(line) < 12:
        return False

    first_part = line[:10]

    if first_part[2] == "/" and first_part[5] == "/":
        return True

    return False
```

```python
with open(file_name, "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total lines:", len(lines))
```

```python
line.split(" - ", 1)
```

```python
messages = []

system_count = 0
media_count = 0
deleted_count = 0

current_message = None


def is_new_message(line):
    line = line.strip()

    # Format: DD/MM/YYYY, HH:MM
    if len(line) >= 17:
        if line[2] == "/" and line[5] == "/" and line[10] == ",":
            return True

    # Format: DD/MM/YY, HH:MM
    if len(line) >= 15:
        if line[2] == "/" and line[5] == "/" and line[8] == ",":
            return True

    return False


for line in lines:

    line = line.rstrip("\n")

    # Check whether this line starts a new WhatsApp message
    if is_new_message(line):

        # Save previous message before starting a new one
        if current_message is not None:
            messages.append(current_message)

        parts = line.split(" - ", 1)

        if len(parts) < 2:
            current_message = None
            continue

        timestamp = parts[0]
        rest = parts[1]

        # Check whether it is a normal message with a sender
        sender_parts = rest.split(": ", 1)

        if len(sender_parts) == 2:

            sender = sender_parts[0]
            text = sender_parts[1].strip()

            # Deleted message
            if text.lower() == "this message was deleted":
                deleted_count += 1
                current_message = None
                continue

            # Media message
            if text == "<Media omitted>":
                media_count += 1

            current_message = {
                "timestamp": timestamp,
                "sender": sender,
                "text": text
            }

        else:
            # System message
            system_count += 1
            current_message = None

    else:
        # This is a continuation of the previous message
        if current_message is not None and line.strip() != "":
            current_message["text"] += " " + line.strip()


# Save the final message
if current_message is not None:
    messages.append(current_message)


# Get unique participants
participants = set()

for message in messages:
    participants.add(message["sender"])


print("Successfully parsed", len(messages), "messages")
print("Participants:", len(participants))
print("System messages:", system_count)
print("Media messages:", media_count)
print("Deleted messages:", deleted_count)
```

```python
stop_words = {
    "the", "is", "a", "an", "and", "or", "to", "of", "in",
    "on", "for", "with", "this", "that", "it", "i", "you",
    "me", "my", "we", "are", "was", "be", "but", "so",
    "if", "at", "as", "from", "have", "has", "had"
}

print("Number of stop words:", len(stop_words))
```

```python
word_frequency = {}

for message in messages:

    text = message["text"]

    # Ignore media messages
    if text == "<Media omitted>":
        continue

    words = text.lower().split()

    for word in words:

        # Remove a few common punctuation marks
        word = word.strip(".,!?;:()[]{}\"'|&")

        if word == "":
            continue

        if word in stop_words:
            continue

        if word not in word_frequency:
            word_frequency[word] = 0

        word_frequency[word] += 1


print("Unique words found:", len(word_frequency))
```

```python
def get_shading(value, maximum):
    if value == 0:
        return " "

    if value <= maximum * 0.25:
        return "░"

    if value <= maximum * 0.50:
        return "▒"

    if value <= maximum * 0.75:
        return "▓"

    return "█"
```

## Feature 6 — Response Times & Silent Streaks

```python
messages_with_datetime = []

for message in messages:

    timestamp = datetime.strptime(
        message["timestamp"],
        "%d/%m/%Y, %H:%M"
    )

    messages_with_datetime.append({
        "timestamp": timestamp,
        "sender": message["sender"],
        "text": message["text"]
    })


messages_with_datetime.sort(key=lambda message: message["timestamp"])

print("Messages sorted by timestamp:", len(messages_with_datetime))
```

```python
response_times = {}

for i in range(len(messages_with_datetime) - 1):

    current = messages_with_datetime[i]
    next_message = messages_with_datetime[i + 1]

    # Only count when a different person sends the next message
    if current["sender"] != next_message["sender"]:

        gap = next_message["timestamp"] - current["timestamp"]

        sender = next_message["sender"]

        if sender not in response_times:
            response_times[sender] = []

        response_times[sender].append(gap.total_seconds() / 60)


print("Response times calculated for", len(response_times), "people.")
```

```python
average_response_times = {}

for person in response_times:

    total_time = sum(response_times[person])
    number_of_responses = len(response_times[person])

    average_response_times[person] = (
        total_time / number_of_responses
    )


print("--- AVERAGE RESPONSE TIMES ---")

for person in average_response_times:

    print(
        person,
        ":",
        round(average_response_times[person], 2),
        "minutes"
    )
```

```python
longest_gap = timedelta(0)
longest_gap_start = None
longest_gap_end = None

for i in range(len(messages_with_datetime) - 1):

    current_time = messages_with_datetime[i]["timestamp"]
    next_time = messages_with_datetime[i + 1]["timestamp"]

    gap = next_time - current_time

    if gap > longest_gap:
        longest_gap = gap
        longest_gap_start = current_time
        longest_gap_end = next_time


print("--- LONGEST SILENT STREAK ---")
print("Duration:", longest_gap)
print("Started:", longest_gap_start)
print("Ended:", longest_gap_end)
```

```python
total_hours = longest_gap.total_seconds() / 3600
total_days = total_hours / 24

print("Longest silent streak:")
print(round(total_days, 2), "days")
print(round(total_hours, 2), "hours")
```

## Feature 7 — Personality Archetypes

```python
def spammer_score(person):
    score = 0

    # High number of messages increases spammer score
    if message_counts[person] > 100:
        score += 3
    elif message_counts[person] > 50:
        score += 2
    elif message_counts[person] > 20:
        score += 1

    # Short average messages also increase the score
    if person in word_counts and message_counts[person] > 0:
        average_length = word_counts[person] / message_counts[person]

        if average_length < 4:
            score += 2
        elif average_length < 7:
            score += 1

    return score
```

```python
def group_mom_score(person):
    score = 0

    # Being active at many different hours
    person_index = participants.index(person)
    active_hours = np.count_nonzero(activity_matrix[person_index])

    if active_hours >= 15:
        score += 3
    elif active_hours >= 10:
        score += 2
    elif active_hours >= 6:
        score += 1

    # Regularly sending messages also increases the score
    if message_counts[person] > 50:
        score += 2
    elif message_counts[person] > 20:
        score += 1

    return score
```

```python
def night_owl_score(person):
    score = 0

    person_index = participants.index(person)

    # Messages sent between 11 PM and 5 AM
    night_messages = (
        activity_matrix[person_index, 23]
        + activity_matrix[person_index, 0]
        + activity_matrix[person_index, 1]
        + activity_matrix[person_index, 2]
        + activity_matrix[person_index, 3]
        + activity_matrix[person_index, 4]
        + activity_matrix[person_index, 5]
    )

    total_messages = message_counts[person]

    if total_messages > 0:
        night_percentage = (night_messages / total_messages) * 100

        if night_percentage >= 40:
            score += 4
        elif night_percentage >= 25:
            score += 3
        elif night_percentage >= 15:
            score += 2
        elif night_percentage >= 5:
            score += 1

    return score
```

```python
def storyteller_score(person):
    score = 0

    if person in word_counts and message_counts[person] > 0:
        average_length = word_counts[person] / message_counts[person]

        if average_length >= 25:
            score += 4
        elif average_length >= 15:
            score += 3
        elif average_length >= 10:
            score += 2
        elif average_length >= 6:
            score += 1

    # A reasonable number of messages helps distinguish
    # a storyteller from someone who sent only one long message.
    if message_counts[person] >= 20:
        score += 1

    return score
```

```python
def drama_queen_score(person):
    score = 0

    person_message_count = 0
    exclamation_count = 0
    question_count = 0

    for message in messages:

        if message["sender"] == person:

            person_message_count += 1

            text = message["text"]

            exclamation_count += text.count("!")
            question_count += text.count("?")

    if person_message_count > 0:

        expressive_marks = exclamation_count + question_count

        marks_per_message = (
            expressive_marks / person_message_count
        )

        if marks_per_message >= 1.5:
            score += 4
        elif marks_per_message >= 1:
            score += 3
        elif marks_per_message >= 0.5:
            score += 2
        elif marks_per_message > 0:
            score += 1

    # Higher activity gives a small additional score
    if person_message_count >= 100:
        score += 2
    elif person_message_count >= 50:
        score += 1

    return score
```

```python
def ghost_score(person):
    score = 0

    person_messages = message_counts[person]

    if person_messages <= 5:
        score += 4
    elif person_messages <= 10:
        score += 3
    elif person_messages <= 20:
        score += 2
    elif person_messages <= 30:
        score += 1

    # Very few active hours also support the ghost score
    person_index = participants.index(person)
    active_hours = np.count_nonzero(activity_matrix[person_index])

    if active_hours <= 2:
        score += 2
    elif active_hours <= 5:
        score += 1

    return score
```

```python
archetype_scores = {}

for person in participants:

    scores = {
        "THE SPAMMER": spammer_score(person),
        "THE GROUP MOM": group_mom_score(person),
        "THE NIGHT OWL": night_owl_score(person),
        "THE STORYTELLER": storyteller_score(person),
        "THE DRAMA QUEEN": drama_queen_score(person),
        "THE GHOST": ghost_score(person)
    }

    archetype_scores[person] = scores


print("Archetype scores calculated for", len(archetype_scores), "people.")
```

```python
person_archetypes = {}

for person in archetype_scores:

    scores = archetype_scores[person]

    best_archetype = max(scores, key=scores.get)

    person_archetypes[person] = best_archetype


print("--- PERSONALITY ARCHETYPES ---")

for person in person_archetypes:

    print(person, "→", person_archetypes[person])
```

```python
person_silent_streaks = {}

for person in participants:

    person_messages = []

    for message in messages_with_datetime:
        if message["sender"] == person:
            person_messages.append(message["timestamp"])

    if len(person_messages) <= 1:
        person_silent_streaks[person] = timedelta(0)
        continue

    longest_silence = timedelta(0)

    for i in range(len(person_messages) - 1):
        gap = person_messages[i + 1] - person_messages[i]

        if gap > longest_silence:
            longest_silence = gap

    person_silent_streaks[person] = longest_silence


print("--- LONGEST SILENT STREAK PER PERSON ---")

for person in person_silent_streaks:
    silence = person_silent_streaks[person]
    print(person, ":", round(silence.total_seconds() / 86400, 2), "days")
```

```python
def ghost_score(person):
    score = 0

    person_messages = message_counts[person]

    # Low total activity is the main Ghost signal
    if person_messages <= 5:
        score += 4
    elif person_messages <= 10:
        score += 3
    elif person_messages <= 20:
        score += 2
    elif person_messages <= 30:
        score += 1

    # Very few active hours support the Ghost score
    person_index = participants.index(person)
    active_hours = np.count_nonzero(activity_matrix[person_index])

    if active_hours <= 2:
        score += 2
    elif active_hours <= 5:
        score += 1

    # Long silent streak is only a supporting factor
    silent_days = (
        person_silent_streaks[person].total_seconds() / 86400
    )

    if silent_days >= 10:
        score += 2
    elif silent_days >= 7:
        score += 1

    return score
```

```python
archetype_scores = {}

for person in participants:

    scores = {
        "THE SPAMMER": spammer_score(person),
        "THE GROUP MOM": group_mom_score(person),
        "THE NIGHT OWL": night_owl_score(person),
        "THE STORYTELLER": storyteller_score(person),
        "THE DRAMA QUEEN": drama_queen_score(person),
        "THE GHOST": ghost_score(person)
    }

    archetype_scores[person] = scores


person_archetypes = {}

for person in archetype_scores:
    scores = archetype_scores[person]
    person_archetypes[person] = max(scores, key=scores.get)


print("--- PERSONALITY ARCHETYPES ---")

for person, archetype in person_archetypes.items():
    print(person, "→", archetype)
```

```python
def get_shading(value, maximum):

    if value == 0:
        return " "

    if maximum == 0:
        return " "

    percentage = value / maximum

    if percentage < 0.25:
        return "░"

    elif percentage < 0.50:
        return "▒"

    elif percentage < 0.75:
        return "▓"

    else:
        return "█"
```

## Feature 6 — Personality Archetypes

```python
def spammer_score(person):

    consecutive_bursts = []
    current_burst = 0

    for message in hostel_messages_with_datetime:

        if message["sender"] == person:
            current_burst += 1

        else:
            if current_burst > 0:
                consecutive_bursts.append(current_burst)

            current_burst = 0

    if current_burst > 0:
        consecutive_bursts.append(current_burst)

    if len(consecutive_bursts) == 0:
        return 0

    average_burst = sum(consecutive_bursts) / len(consecutive_bursts)

    if average_burst > 3:
        return 4
    elif average_burst > 2:
        return 2
    else:
        return 0
```

```python
def group_mom_score(person):

    caring_keywords = [
        "okay",
        "safe",
        "eat",
        "sleep",
        "take care",
        "are you",
        "please",
        "reminder",
        "drink water",
        "don't forget"
    ]

    score = 0

    for message in hostel_messages:

        if message["sender"] != person:
            continue

        text = message["text"].lower()

        for keyword in caring_keywords:

            if keyword in text:
                score += 1

    return score
```

```python
def night_owl_score(person):

    total_messages = 0
    night_messages = 0

    for message in hostel_messages_with_datetime:

        if message["sender"] != person:
            continue

        total_messages += 1

        hour = message["timestamp"].hour

        if hour >= 23 or hour <= 4:
            night_messages += 1

    if total_messages == 0:
        return 0

    percentage = (night_messages / total_messages) * 100

    if percentage > 60:
        return 4
    else:
        return 0
```

```python
def storyteller_score(person):

    total_words = 0
    total_messages = 0

    for message in hostel_messages:

        if message["sender"] == person:

            total_messages += 1
            total_words += len(message["text"].split())

    if total_messages == 0:
        return 0

    average_words = total_words / total_messages

    if average_words > 30:
        return 4
    else:
        return 0
```

```python
def drama_queen_score(person):

    total_messages = 0
    dramatic_messages = 0

    for message in hostel_messages:

        if message["sender"] != person:
            continue

        text = message["text"].strip()

        if len(text) < 3:
            continue

        total_messages += 1

        words = text.split()

        uppercase_message = False

        if len(words) > 0:

            uppercase_words = 0

            for word in words:

                if any(character.isalpha() for character in word):
                    if word.isupper():
                        uppercase_words += 1

            if uppercase_words == len(words):
                uppercase_message = True

        if uppercase_message or text.count("!") >= 2:
            dramatic_messages += 1

    if total_messages == 0:
        return 0

    percentage = (dramatic_messages / total_messages) * 100

    if percentage > 30:
        return 4
    else:
        return 0
```

```python
def ghost_score(person):

    all_dates = set()
    person_dates = set()

    for message in hostel_messages_with_datetime:

        date = message["timestamp"].date()

        all_dates.add(date)

        if message["sender"] == person:
            person_dates.add(date)

    if len(all_dates) == 0:
        return 0

    silent_days = len(all_dates - person_dates)

    silent_percentage = (
        silent_days / len(all_dates)
    ) * 100

    if silent_percentage > 60:
        return 4
    else:
        return 0
```

## Feature 7 — Archetype

```python
# Feature 7 - Personality Archetype System

def hostel_average_burst(person):
    bursts = []
    current_burst = 0

    for message in hostel_messages_with_datetime:
        if message["sender"] == person:
            current_burst += 1
        else:
            if current_burst > 0:
                bursts.append(current_burst)
            current_burst = 0

    if current_burst > 0:
        bursts.append(current_burst)

    if len(bursts) == 0:
        return 0

    return sum(bursts) / len(bursts)


def hostel_group_mom_keyword_count(person):
    caring_keywords = [
        "okay", "safe", "eat", "sleep", "take care",
        "are you", "please", "reminder", "drink water",
        "don't forget"
    ]

    count = 0

    for message in hostel_messages:
        if message["sender"] == person:
            text = message["text"].lower()

            for keyword in caring_keywords:
                if keyword in text:
                    count += 1

    return count


def hostel_night_percentage(person):
    total = 0
    night = 0

    for message in hostel_messages_with_datetime:
        if message["sender"] == person:
            total += 1
            hour = message["timestamp"].hour

            if hour >= 23 or hour <= 4:
                night += 1

    if total == 0:
        return 0

    return (night / total) * 100


def hostel_average_words(person):
    total_words = 0
    total_messages = 0

    for message in hostel_messages:
        if message["sender"] == person:
            total_messages += 1
            total_words += len(message["text"].split())

    if total_messages == 0:
        return 0

    return total_words / total_messages


def hostel_drama_percentage(person):
    total = 0
    dramatic = 0

    for message in hostel_messages:
        if message["sender"] == person:
            total += 1
            text = message["text"]

            letters = ""

            for character in text:
                if character.isalpha():
                    letters += character

            all_caps = (
                len(letters) >= 3 and
                letters.upper() == letters
            )

            two_exclamations = text.count("!") >= 2

            if all_caps or two_exclamations:
                dramatic += 1

    if total == 0:
        return 0

    return (dramatic / total) * 100


def hostel_silent_percentage(person):
    all_dates = set()
    person_dates = set()

    for message in hostel_messages_with_datetime:
        date = message["timestamp"].date()
        all_dates.add(date)

        if message["sender"] == person:
            person_dates.add(date)

    if len(all_dates) == 0:
        return 0

    silent_days = len(all_dates) - len(person_dates)

    return (silent_days / len(all_dates)) * 100


# Find the single Group Mom
hostel_group_mom_counts = {}

for person in hostel_participants:
    hostel_group_mom_counts[person] = hostel_group_mom_keyword_count(person)

hostel_group_mom_person = max(
    hostel_group_mom_counts,
    key=hostel_group_mom_counts.get
)


# Assign the six archetypes
hostel_person_archetypes = {}

for person in hostel_participants:

    # Group Mom
    if person == hostel_group_mom_person:
        archetype = "GROUP MOM"

    # Spammer
    elif hostel_average_burst(person) > 3:
        archetype = "SPAMMER"

    # Night Owl
    elif hostel_night_percentage(person) > 60:
        archetype = "NIGHT OWL"

    # Storyteller
    elif hostel_average_words(person) > 30:
        archetype = "STORYTELLER"

    # Drama Queen
    elif hostel_drama_percentage(person) > 30:
        archetype = "DRAMA QUEEN"

    # Ghost
    elif hostel_silent_percentage(person) > 60:
        archetype = "GHOST"

    else:
        archetype = "UNCLASSIFIED"

    hostel_person_archetypes[person] = archetype


print("--- FEATURE 7: PERSONALITY ARCHETYPES ---")

for person in hostel_participants:
    print(person, "->", hostel_person_archetypes[person])
```
